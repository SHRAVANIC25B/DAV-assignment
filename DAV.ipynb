```python
#data processing:handling missing data using imputation(mean,median,forward fill) and encoding categorical variables
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Using a dataset that exists on the Colab environment
df = pd.read_csv('/content/sample_data/california_housing_train.csv')

print("Original Dataset:")
print(df.head())
print("\nShape of dataset:")
print(df.shape)
print("\nColumn names:")
print(df.columns)
print("\nDataset information:")
print(df.info())
print("\nStatistical summary:")
print(df.describe())
print("\nMissing Values Before Preprocessing:")
print(df.isnull().sum())

# Demonstrate Mean Imputation on 'total_bedrooms'
df_mean = df.copy()
mean_bedrooms = df_mean["total_bedrooms"].mean()
df_mean["total_bedrooms"] = df_mean["total_bedrooms"].fillna(
    mean_bedrooms
)
print("\nMean Total Bedrooms:", mean_bedrooms)
print("Missing total_bedrooms values after Mean Imputation:")
print(df_mean["total_bedrooms"].isnull().sum())

# Demonstrate Median Imputation on 'housing_median_age'
df_median = df.copy()
median_age = df_median["housing_median_age"].median()
df_median["housing_median_age"] = df_median["housing_median_age"].fillna(
    median_age
)
print("\nMedian Housing Age:", median_age)
print("Missing housing_median_age values after Median Imputation:")
print(df_median["housing_median_age"].isnull().sum())

# Demonstrate Forward Fill on 'total_rooms'
df_ffill = df.copy()
df_ffill["total_rooms"] = df_ffill["total_rooms"].ffill()
print("\nMissing total_rooms values after Forward Fill:")
print(df_ffill["total_rooms"].isnull().sum())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading the Iris dataset from an online repository to avoid FileNotFoundError in Colab
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")

# VIEW THE FIRST 5 ROWS
print("First 5 rows of the dataset:")
print(df.head())

# VIEW THE LAST 5 ROWS
print("\nLast 5 rows of the dataset:")
print(df.tail())

# CHECK THE SIZE OF THE DATASET
print("\nShape of the dataset:")
print(df.shape)

# VIEW COLUMN NAMES
print("\nColumn names:")
print(df.columns)

# CHECK DATASET INFORMATION
print("\nDataset information:")
df.info()

# Check missing values
print(df.isnull().sum())

# Remove missing values
df = df.dropna()

# Check missing values again
print(df.isnull().sum())

# Check duplicate rows
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check duplicates again
print(df.duplicated().sum())

# Display statistical summary
print(df.describe())

# Set graph style
sns.set_style("whitegrid")

# ---------------- HISTOGRAM ----------------

# Create graph area
plt.figure(figsize=(8, 5))

# Create histogram
sns.histplot(data=df, x="sepal_length", bins=10)

# Add title
plt.title("Distribution of Sepal Length")

# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Number of Flowers")

# Display graph
plt.show()

# ---------------- BOX PLOT ----------------

# Create graph area
plt.figure(figsize=(8, 5))

# Create box plot
sns.boxplot(data=df, x="species", y="sepal_length")

# Add title
plt.title("Sepal Length by Species")

# Add X-axis label
plt.xlabel("Species")

# Add Y-axis label
plt.ylabel("Sepal Length (cm)")

# Display graph
plt.show()

# ---------------- SCATTER PLOT ----------------

# Create graph area
plt.figure(figsize=(8, 5))

# Create scatter plot
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)

# Add title
plt.title("Sepal Length vs Petal Length")

# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Petal Length (cm)")

# Display graph
plt.show()

# ---------------- PAIR PLOT ----------------

# Create pair plot
sns.pairplot(df, hue="species")

# Display graph
plt.show()

# ---------------- CORRELATION ----------------

# Select numerical columns
numeric_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

# Calculate correlation
correlation = df[numeric_columns].corr()

# Display correlation matrix
print(correlation)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
import pandas as pd
import numpy as np

# Programmatically generating a synthetic dataset to replace the missing local file
np.random.seed(42)
n_students = 250

synthetic_data = {
    'student_id': [f"STU_{1000 + i}" for i in range(n_students)],
    'math_score': np.random.randint(40, 100, size=n_students),
    'reading_score': np.random.randint(45, 100, size=n_students),
    'writing_score': np.random.randint(40, 100, size=n_students),
    'grade': np.random.choice(['A', 'B', 'C', 'D', 'F'], size=n_students, p=[0.2, 0.3, 0.3, 0.15, 0.05])
}

df = pd.DataFrame(synthetic_data)

print("Dataset loaded successfully!")

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nShape of the dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nDescriptive Statistics:")
print(df.describe())

print("\nStatistics including categorical columns:")
print(df.describe(include='all'))

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

print("\nNumber of duplicate student IDs:")
print(df['student_id'].duplicated().sum())

print("\nNumber of students in each grade:")
print(df['grade'].value_counts())

grade_info = pd.DataFrame({
    'grade': ['A', 'B', 'C', 'D', 'F'],
    'grade_description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ]
})

merged = pd.merge(
    df,
    grade_info,
    on='grade',
    how='left'
)

print("\nMerged DataFrame:")
print(merged.head())

grade_details = pd.DataFrame({
    'description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ],
    'minimum_score': [90, 75, 60, 50, 0],
    'maximum_score': [100, 89, 74, 59, 49]
})

print("\nGrade Details:")
print(grade_details)

df1 = df.iloc[:100]
df2 = df.iloc[100:200]

combined = pd.concat([df1, df2])

print("\nConcatenated DataFrame:")
print(combined.shape)
```
